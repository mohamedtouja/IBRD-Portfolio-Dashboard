# Data Analysis Notebook

This template loads data and creates a quick exploratory summary.

```python
import pandas as pd
import matplotlib.pyplot as plt
print("pandas and matplotlib ready")
```

```python
df = pd.DataFrame({"x": [1, 2, 3, 4], "y": [10, 15, 13, 16]})
print(df.head())
```

```python
df.plot(kind="line", x="x", y="y")
plt.title("Example trend")
plt.show()
```
